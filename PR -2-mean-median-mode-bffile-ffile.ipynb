```python
#Level 2: User-Defined Functions for missing value Handlings

#step 1 : import lib
import numpy as np
import pandas as pd
```

```python
#step 2 : Create Sample Dataset with Missing Values
data={
    'Age':[25,30,np.nan,45,50],
    'Salary':[50000,np.nan,60000,65000,np.nan],
    'Experience':[1,3,5,np.nan,10]

}
df= pd.DataFrame(data)
df
```

```python
#User- Defined Functions for missing values handling
#exercise 1: Remove Rows with Missing Values

def remove_missing_rows(dataframe):
  return dataframe.dropna()
```

```python
#Usage:
df_no_missing= remove_missing_rows(df)
df_no_missing
```

```python
#Exercise 2: Mean Imptation
def mean_imputation(dataframe,column_name):
  mean_value = dataframe[column_name].mean()
  dataframe[column_name] = dataframe[column_name].fillna(mean_value)
  return dataframe
```

```python
#Usage:
df_mean = mean_imputation(df.copy(),'Age')
df_mean
```

```python
#Exercise 3 : Meadian Imputation
def median_imputation(dataframe ,column_name):
  median_value = dataframe[column_name].median()
  dataframe[column_name] = dataframe[column_name].fillna(median_value)
  return dataframe
```

```python
#Usage:
df_median = median_imputation(df.copy(),'Salary')
df_median
```

```python
#Exercise 4 : Mode Imputation
def mode_imputation(dataframe ,column_name):
  mode_value = dataframe[column_name].mode()[0]
  dataframe[column_name] = dataframe[column_name].fillna(mode_value)
  return dataframe
```

```python
#Usage:
df_mode = mode_imputation(df.copy(),'Experience')
df_mode
```

```python
#Exercise 5 : Forward Fill (ffill)
def forward_fill(dataframe ,column_name):
  dataframe[column_name] = dataframe[column_name].fillna(method = 'ffill')
  return dataframe
```

```python
#usage
df_ffill = forward_fill(df.copy(),'Salary')
df_ffill
```

```python
#Exercise 6:Backward Fill (bfill)
def backward_fill(dataframe ,column_name):
  dataframe[column_name] = dataframe[column_name].fillna(method = 'bfill')
  return dataframe
```

```python
#usage
df_bfill = backward_fill(df.copy(),'Experience')
df_bfill
```
