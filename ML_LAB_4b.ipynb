Data Normalization concepts - Without Lib using math

```python
import pandas as pd
import numpy as np
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

columns = ['SepalLength', 'SepalWidth','PetalLength','PetalWidth', 'Species']
data = pd.read_csv(url,header=None , names=columns)

x=data.iloc[:,:-1]
```

```python
#Define Custom Min-Max Normalization Function
def min_max_normalization(column):
  min_val = column.min()
  max_val = column.max()
  return (column - min_val) / (max_val - min_val)
```

```python
#Apply Min-Max Normalization

x_minmax = x.apply(min_max_normalization)
x_minmax.head()
```

```python
#define custom z- score normalization functions
def z_score_normalization(column):
  mean_val = column.mean()
  std_val = column.std()
  return (column - mean_val) / std_val
```

```python
#Apply Z-score normalization
x_zscore = x.apply(z_score_normalization)
x_zscore.head()
```

```python
#define Custom Robust scaling function
def robust_scaling(column):
  median_val = column.median()
  q1 = column.quantile(0.25)
  q3 = column.quantile(0.75)
  return (column - median_val) / (q3 - q1)
```

```python
#Apply Robust Scaling
x_robust = x.apply(robust_scaling)
x_robust.head()
```
