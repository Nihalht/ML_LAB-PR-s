Data Normalization concepts

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler,StandardScaler,RobustScaler
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

columns = ['SepalLength', 'SepalWidth','PetalLength','PetalWidth', 'Species']
data = pd.read_csv(url,header=None , names=columns)
data.head()
```

```python
#Separate Features and target

x= data.iloc[:,:-1] #Numerical Features
y= data['Species'] #Target Variable

```

```python
#Min-Max Normalization
#Scales data to a fixed range, usually[0, 1].

minmax_scaler = MinMaxScaler()
x_minmax = minmax_scaler.fit_transform(x)

pd.DataFrame(x_minmax,columns=x.columns).head()
```

```python
#Standardization(Z-score Normalization)
#Centers the data around zero with unit varience

standard_scaler = StandardScaler()
x_standard = standard_scaler.fit_transform(x)

pd.DataFrame(x_standard,columns=x.columns).head()
```

```python
#Robust Scaling
#Scales the data using statistics that are robust to outliers

robust_scaler = RobustScaler()
x_robust = robust_scaler.fit_transform(x)

pd.DataFrame(x_robust,columns=x.columns).head()
```
