```python
#Experiment 5a: Regression learning
#step1:

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split,learning_curve
from sklearn.metrics import mean_squared_error,r2_score
```

```python
#step 2 load dataset

url="https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
columns = ['SepalLength','SepalWidth','PetalLength','PetalWidth','Species']
data = pd.read_csv(url, header=None, names=columns)
```

```python
x = data[['SepalLength']]
y = data[['PetalLength']]
```

```python
#step 3 : Split data set

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
#Step 4:Implement Simple Linear Regression
model=LinearRegression()
model.fit(x_train,y_train)
```

```python
#step 5 : Evaluate Model Parameters
print("slope (Coefficient):",model.coef_[0])
print("Intercept", model.intercept_)
```

```python
#step 6 :Make predication

y_pred = model.predict(x_test)
```

```python
#step 7 : Evalute  model performance

mse = mean_squared_error(y_test,y_pred)
r2 = r2_score(y_test,y_pred)

print("Mean Squared Error:",mse)
print("R-squared:",r2)
```

```python
#Step 8 : Plot Regression Line

plt.figure()
plt.scatter(x_test,y_test,color='black')
plt.plot(x_test,y_pred,color='blue',linewidth=2)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.title('Simple Linear Regression')
plt.show()
```

```python
#step 9 : plot leaning curves

train_sizes,train_scores,test_scores = learning_curve(model,x,y,cv=5,scoring='r2')
train_scores_mean = np.mean(train_scores,axis=1)
test_scores = np.mean(test_scores,axis=1)
```

```python
plt.figure()
plt.plot(train_sizes,train_scores_mean,'o-',color='blue',label='Training Score')
plt.plot(train_sizes,test_scores,'o-',color='green',label='Cross-Validation Score')
plt.xlabel('Training Examples')
plt.ylabel('R-squared Score')
plt.title('Learning Curves')
plt.legend()
plt.show()
```
